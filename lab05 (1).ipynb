### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```sage
groups_data = [
    ("Z_400", AbelianGroup([400])),
    ("Z_16 x Z_25", AbelianGroup([16, 25])),
    ("Z_20 x Z_20", AbelianGroup([20, 20])),
    ("Z_4 x Z_100", AbelianGroup([4, 100])),
    ("Z_2 x Z_10 x Z_20", AbelianGroup([2, 10, 20])),
    ("Z_505*", IntegerModRing(505).unit_group())
]

for label, grp in groups_data:
    print("\nГрупа:", label)
    inv_divs = grp.invariants()
    print("1. Елементарні дільники:", inv_divs)
    
    p_groups = []
    for d in inv_divs:
        for p, k in factor(d):
            p_groups.append(p^k)
    p_groups.sort()
    print("2. Розклад у добуток циклічних p-груп:", p_groups)
    
    ord_counts = {}
    for el in grp:
        o = el.order()
        ord_counts[o] = ord_counts.get(o, 0) + 1
    print("3. Кількість елементів кожного порядку:", dict(sorted(ord_counts.items())))

print("\n4. Ізоморфні групи:")
print("Z_400 та Z_16 x Z_25")
print("Z_4 x Z_100 та Z_505*")
print("Z_20 x Z_20 (не ізоморфна іншим)")
print("Z_2 x Z_10 x Z_20 (не ізоморфна іншим)")
```

### 2. Одна абелева група порядку $n$
1. Реалізуйте функцію, яка для заданого натурального числа $n$ обчислює кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```sage
def count_abelian_groups(n):
    total = 1
    for p, exp in factor(n):
        total *= Partitions(exp).cardinality()
    return total

unique_n = []
curr = 1
while len(unique_n) < 100:
    if count_abelian_groups(curr) == 1:
        unique_n.append(curr)
    curr += 1

print("2. Перші 100 чисел n з єдиною абелевою групою та їх факторизація:")
for idx, val in enumerate(unique_n, 1):
    print(f"{idx}. n = {val} | розклад: {factor(val)}")

print("\n3. Гіпотеза:")
print("Існує рівно одна абелева група порядку n тоді і лише тоді, коли n є вільним від квадратів (усі показники степенів простих множників дорівнюють 1, або n = 1).")
```

### 3. Дільники нуля та одиниці в скінченних кільцях
Розгляньте скінченні кільця $\mathbb{Z}_{30}$, $\mathbb{Z}_{12}\times\mathbb{Z}_{15}$, $M_3(\mathbb{Z}_3)$ та для кожного з них виконайте наступні кроки:
1. Знайдіть дільники нуля та їх кількість.
2. Знайдіть дільники одиниці та їх кількість.
3. Порівняйте отримані результати та зробіть висновок про взаємозв'язок між дільниками нуля та дільниками одиниці в скінченних кільцях.


```sage
R1 = IntegerModRing(30)
zd1 = [x for x in R1 if x != 0 and not x.is_unit()]
un1 = [x for x in R1 if x.is_unit()]
print("Кільце Z_30 (порядок =", R1.cardinality(), "):")
print("1. Дільники нуля (кількість =", len(zd1), "):", zd1)
print("2. Дільники одиниці (кількість =", len(un1), "):", un1)

R2 = cartesian_product([IntegerModRing(12), IntegerModRing(15)])
zd2 = []
un2 = []
for x in R2:
    if x == R2.zero():
        continue
    if x[0].is_unit() and x[1].is_unit():
        un2.append(x)
    else:
        zd2.append(x)
print("\nКільце Z_12 x Z_15 (порядок =", R2.cardinality(), "):")
print("1. Дільники нуля: кількість =", len(zd2))
print("2. Дільники одиниці: кількість =", len(un2))

R3 = MatrixSpace(GF(3), 3, 3)
total_m = R3.cardinality()
un3_count = GL(3, GF(3)).order()
zd3_count = total_m - un3_count - 1
print("\nКільце M_3(Z_3) (порядок =", total_m, "):")
print("1. Дільники нуля: кількість =", zd3_count)
print("2. Дільники одиниці: кількість =", un3_count)

print("\n3. Висновок:")
print("У скінченному кільці R з одиницею кожен ненульовий елемент є або оборотним, або дільником нуля: |R| = |U(R)| + |ZD(R)| + 1.")
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```sage
for name, base_ring in [("Q", QQ), ("Z_4", IntegerModRing(4)), ("Z_5", GF(5))]:
    P.<x> = PolynomialRing(base_ring)
    f = x^4 + x + 2
    I = P.ideal(f)
    Q_ring = P.quotient(I, 'a')
    
    print("\nБазове кільце R =", name)
    print("1. Ідеал I =", I)
    
    if base_ring == QQ:
        print("2. Порядок факторкільця R/I: нескінченний")
        is_fld = f.is_irreducible()
        print("3. Чи є R/I полем:", is_fld)
        print("4. Дільники нуля у R/I:", "відсутні" if is_fld else "існують")
        print("5. Розклад f(x) над R:", f.factor())
        x_bar = Q_ring.gen()
        print("6. Елемент x + I оборотний:", x_bar.is_unit())
        if x_bar.is_unit():
            print("   Обернений до x + I:", x_bar^(-1))
    elif base_ring == GF(5):
        print("2. Порядок факторкільця R/I:", Q_ring.cardinality())
        is_fld = Q_ring.is_field()
        print("3. Чи є R/I полем:", is_fld)
        z_divs = [el for el in Q_ring if el != 0 and not el.is_unit()]
        print("4. Дільники нуля у R/I (кількість =", len(z_divs), "):", z_divs[:10])
        print("5. Розклад f(x) над R:", f.factor())
        x_bar = Q_ring.gen()
        print("6. Елемент x + I оборотний:", x_bar.is_unit())
        if x_bar.is_unit():
            print("   Обернений до x + I:", x_bar^(-1))
    else:
        print("2. Порядок факторкільця R/I:", Q_ring.cardinality())
        print("3. Чи є R/I полем: False")
        elems = list(Q_ring)
        z_divs = []
        for el1 in elems:
            if el1 == 0:
                continue
            for el2 in elems:
                if el2 != 0 and el1 * el2 == 0:
                    z_divs.append(el1)
                    break
        print("4. Дільники нуля у R/I (кількість =", len(z_divs), "):", z_divs[:10])
        roots_z4 = [r for r in base_ring if f(r) == 0]
        print("5. Корені f(x) у Z_4:", roots_z4)
        x_bar = Q_ring.gen()
        inv_x = [el for el in elems if x_bar * el == 1]
        print("6. Елемент x + I оборотний:", len(inv_x) > 0)
        if len(inv_x) > 0:
            print("   Обернений до x + I:", inv_x[0])
```

### 5*. Корінь з $-1$ за модулем $p$

Виконайте наступні кроки і на їх основі сформулюйте гіпотезу: для яких простих чисел $p$ фактор-кільце $\mathbb{Z}_p[x]/(x^2+1)$ є полем, тобто многочлен $x^2+1$ не має коренів у $\mathbb{Z}_p$.
1. Знайдіть перші 100 простих чисел $p$, для яких факторкільце $\mathbb{Z}_p[x]/(x^2+1)$ є полем.
2. Визначте остачі цих простих чисел при діленні на $m=2,3,\ldots,7$ та проаналізуйте отримані результати.


```sage
primes_field = []
p = 2
while len(primes_field) < 100:
    P.<x> = PolynomialRing(GF(p))
    if (x^2 + 1).is_irreducible():
        primes_field.append(p)
    p = next_prime(p)

print("1. Перші 100 простих чисел p, для яких Z_p[x]/(x^2+1) є полем:")
print(primes_field)

print("\n2. Остачі цих простих чисел при діленні на m = 2..7:")
for m in range(2, 8):
    rems = sorted(list(set([pr % m for pr in primes_field])))
    print(f"m = {m}: остачі = {rems}")

print("\nГіпотеза:")
print("Факторкільце Z_p[x]/(x^2+1) є полем тоді і лише тоді, коли p ≡ 3 (mod 4).")
```
