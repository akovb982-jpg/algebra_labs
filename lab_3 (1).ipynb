### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
# Ваш код тут:

for n in [100, 173, 530, 746, 911, 1024]:
    G = Integers(n)
    divs = divisors(n)
    ords = [g.additive_order() for g in G]

    print("\nn =", n)
    print("1. Дільники:", divs)
    print("2–3. Порядок d | кількість | phi(d) | збіг")

    for d in divs:
        cnt = ords.count(d)
        print(d, cnt, euler_phi(d), cnt == euler_phi(d))

    print("4. Усі підгрупи (вказано твірний та елементи):")
    subs = []

    for d in divs:
        gen = G(n // d)
        H = [ZZ(j) * gen for j in range(d)]
        subs.append(H)
        print("Порядок", d, "; твірний", gen, "; H =", H)

    print("5. Кількість підгруп:", len(subs))
    print("6. tau(n) =", number_of_divisors(n))
    print("Збіг:", len(subs) == number_of_divisors(n))
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
# Ваш код тут:
for n in [5, 6, 10, 12, 16, 17, 20]:
    D = DihedralGroup(n)
    ords = [g.order() for g in D]
    N = D.order()

    print("\nn =", n)
    print("|D_n| =", N)
    print("1. Кількість елементів порядку 2:", ords.count(2))

    print("2. Дільник d | кількість елементів порядку d")
    for d in divisors(N):
        if d > 2:
            print(d, "|", ords.count(d))

    print("3. Чи існує елемент кожного порядку d < 2n?")
    for d in divisors(N):
        if d < N:
            print("d =", d, "| існує:", d in ords)
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
# Ваш код тут:
N = 1102158078129785185997827239806

print("1. Факторизація n:")
F = factor(N)
print(F)

print("\n2. Порядок групи:")
order = euler_phi(N)
print("phi(n) =", order)

def is_cyclic_unit_group(N, F):
    if N in [1, 2, 4]:
        return True
    if len(F) == 1 and F[0][0] != 2:
        return True
    if len(F) == 2 and F[0][0] == 2 and F[0][1] == 1 and F[1][0] != 2:
        return True
    return False

cyclic = is_cyclic_unit_group(N, F)
print("Група циклічна:", cyclic)

if cyclic:
    print("\n3. Пошук найменшого твірного елемента:")
    a = 2
    while gcd(a, N) != 1 or Mod(a, N).multiplicative_order() != order:
        a += 1
    print("Найменший твірний елемент a =", a)

    print("\n4. Порядок елемента:")
    print("ord(a) =", Mod(a, N).multiplicative_order())

    print("\n5. Випадкові елементи:")
    gs = []
    while len(gs) < 3:
        g = ZZ.random_element(1, N)
        if gcd(g, N) == 1 and g not in gs:
            gs.append(g)
    print(gs)

    print("\n6. Дискретні логарифми:")
    for g in gs:
        k = discrete_log(Mod(g, N), Mod(a, N))
        print("g =", g, "; k =", k)
        print("Перевірка:", power_mod(a, k, N) == g)
else:
    print("Група не циклічна.")
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:
1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?

```sage
# Ваш код тут:

A = matrix(ZZ, [[1, 0, 1],
                [0, 1, 0],
                [0, 0, 1]])

B = matrix(ZZ, [[1, 0, 0],
                [0, 1, 1],
                [0, 0, 1]])

E = identity_matrix(ZZ, 3)

print("A * B = B * A:", A * B == B * A)

u, v, s, t, k = var("u v s t k")

X = matrix(SR, [[1, 0, u],
                [0, 1, v],
                [0, 0, 1]])

Y = matrix(SR, [[1, 0, s],
                [0, 1, t],
                [0, 0, 1]])

N = X - identity_matrix(SR, 3)

print("\nЗагальний вигляд X(u,v):")
show(X)

print("Перевірка N^2 = 0:", N^2 == zero_matrix(SR, 3))

print("\nДобуток X(u,v) * X(s,t):")
show(X * Y)

print("\nФормула X(u,v)^k = I + k*N:")
show(identity_matrix(SR, 3) + k * N)

print("\nПриклади степенів A^j * B^i:")

for j, i in [(-2, 3), (0, 0), (1, -1), (3, 2)]:
    expected = matrix(ZZ, [[1, 0, j],
                           [0, 1, i],
                           [0, 0, 1]])

    result = A^j * B^i

    print("j =", j, "; i =", i,
          "; збіг:", result == expected)
    show(result)

print("\nВИСНОВКИ")
print("1. Група нескінченна: степені A різні.")
print("2. Група абелева: A і B комутують.")
print("3. Порядок одиничної матриці дорівнює 1.")
print("4. Кожен не одиничний елемент має нескінченний порядок.")
print("5. Елементи мають вигляд X(u,v), де u,v — цілі числа.")
print("6. Група не циклічна.")
print("7. Група ізоморфна Z x Z.")
```

### 5*. Циклічність груп $\mathbb{Z}_n^{*}$

Для кожного $n=2, 3, 4, \ldots, 256$ виконайте наступні кроки, представляючи результати у вигляді таблиці:
1. Знайдіть факторизацію числа $n$.
2. Перевірте, чи є група $\mathbb{Z}_n^{*}$ циклічною.
3. Для циклічних груп $\mathbb{Z}_n^{*}$ знайдіть найменше $a \in \mathbb{N}$ таке, що $\overline{a}$ є твірним групи.
4. Сформулюйте гіпотезу: для яких $n \in \mathbb{N}$ група $\mathbb{Z}_n^{*}$ є циклічною.

```sage
rows = []

for n in range(2, 257):
    phi_n = euler_phi(n)
    units = [a for a in range(1, n) if gcd(a, n) == 1]
    exponent = max(Mod(a, n).multiplicative_order() for a in units)
    cyclic = (exponent == phi_n)
    smallest = None
    if cyclic:
        smallest = next(a for a in units if Mod(a, n).multiplicative_order() == phi_n)
    rows.append((n, factor(n), phi_n, cyclic, smallest))

print("n".rjust(4), "факторизація".ljust(24), "phi(n)".rjust(7), "циклічна".ljust(10), "найменший твірний")

for n, f, ph, cyc, g in rows:
    print(str(n).rjust(4), str(f).ljust(24), str(ph).rjust(7), ("так" if cyc else "ні").ljust(10), g if cyc else "-")

cyclic_list = [r[0] for r in rows if r[3]]
print()
print("Кількість n, для яких група циклічна:", len(cyclic_list))
print("Такі n:", cyclic_list)

def predicted(n):
    if n in [2, 4]:
        return True
    if n % 2 == 1:
        return is_prime_power(n)
    return n % 4 == 2 and is_prime_power(n // 2)

print()
print("Гіпотеза: Z_n^* циклічна тоді й лише тоді, коли n = 2, 4, p^k або 2*p^k, де p — непарне просте.")
print("Збігається з обчисленнями для всіх n = 2..256:", all(predicted(n) == cyc for n, f, ph, cyc, g in rows))
```

### 6*. Числа Фібоначчі

Розглянемо підгрупу $H$ групи $GL_2(\mathbb{Z})$, породжену двома матрицями:
$$H = \left\langle \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix}, \begin{pmatrix} 3 & 2 \\ 2 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath: 

1. Чи є група $H$ скінченною?
2. Чи є група $H$ абелевою?
3. Які порядки мають елементи групи $H$?
4. Який вигляд мають елементи групи $H$?
5. Чи є група $H$ циклічною? Якщо так, то знайдіть її твірний елемент.

```sage
P = matrix(ZZ, [[1, -1],
                [-1, 2]])

Q = matrix(ZZ, [[3, 2],
                [2, 1]])

E2 = identity_matrix(ZZ, 2)

print("det P =", P.det(), "; det Q =", Q.det())
print("P * Q = Q * P:", P * Q == Q * P)

print("\nДобуток P * Q:")
show(P * Q)

M = matrix(ZZ, [[1, 1],
                [1, 0]])

print("\nP * Q = M:", P * Q == M)
print("P = M^(-2):", P == M^(-2))
print("Q = M^3:", Q == M^3)
print("M = Q * P (отже H = <M>):", M == Q * P)

def fib_int(n):
    if n >= 0:
        return fibonacci(n)
    return (-1)^(-n + 1) * fibonacci(-n)

print("\nЗагальний вигляд M^n = [[F(n+1), F(n)], [F(n), F(n-1)]]:")
print("Перевірка для n від -15 до 15:",
      all(M^n == matrix(ZZ, [[fib_int(n + 1), fib_int(n)],
                             [fib_int(n), fib_int(n - 1)]])
          for n in range(-15, 16)))

print("\nХарактеристичний многочлен M:", M.charpoly())
print("Власні числа M:", M.eigenvalues())

print("\nПриклади степенів P^j * Q^i = M^(3i - 2j):")

for j, i in [(-2, 3), (0, 0), (1, -1), (3, 2)]:
    result = P^j * Q^i
    expected = M^(3*i - 2*j)

    print("j =", j, "; i =", i, "; показник =", 3*i - 2*j,
          "; збіг:", result == expected)
    show(result)

print("\nМатриця M^n = E лише при n = 0 (перевірка n від -50 до 50):",
      [n for n in range(-50, 51) if M^n == E2])

print("\nЗростання елементів M^n:")
for n in [1, 5, 10, 20, 40]:
    print("n =", n, "; max |елемент| =", max(abs(x) for x in (M^n).list()))

print("\nВИСНОВКИ")
print("1. Група нескінченна: M^n мають різні елементи (числа Фібоначчі зростають).")
print("2. Група абелева: P і Q комутують (обидві — степені M).")
print("3. Порядок одиничної матриці дорівнює 1, усі інші елементи мають нескінченний порядок.")
print("4. Елементи мають вигляд M^n = [[F(n+1), F(n)], [F(n), F(n-1)]], n — ціле.")
print("5. Група циклічна: H = <M>, твірний елемент M = [[1, 1], [1, 0]] (M = Q*P).")
print("   H ізоморфна (Z, +).")
```
