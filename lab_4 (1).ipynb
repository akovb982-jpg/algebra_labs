### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
S4 = SymmetricGroup(4)
subgroups = [
    ("H1 = <(1,2,3,4)>", S4.subgroup([S4("(1,2,3,4)")])),
    ("H2 = <(1,3), (1,2,3,4)>", S4.subgroup([S4("(1,3)"), S4("(1,2,3,4)")])),
    ("H3 = K4", S4.subgroup([S4("(1,2)(3,4)"), S4("(1,3)(2,4)")])),
]

for title, sub in subgroups:
    print()
    print("Підгрупа", title)
    print("Елементи:", list(sub))
    print("Порядок |H| =", sub.order(), "; індекс [S4 : H] =", S4.order() // sub.order())

    print("\nЛіві класи суміжності aH:")
    for coset in S4.cosets(sub, side="left"):
        print("  ", coset)

    print("\nПраві класи суміжності Ha:")
    for coset in S4.cosets(sub, side="right"):
        print("  ", coset)

    is_norm = sub.is_normal(S4)
    print("\nH нормальна в S4:", is_norm)

    if is_norm:
        quot = S4.quotient(sub)
        print("Порядок G/H:", quot.order())
        print("Структура G/H:", quot.structure_description())
        print("G/H ≅ S3:", quot.is_isomorphic(SymmetricGroup(3)))
        print("Таблиця Келі G/H:")
        print(quot.cayley_table(names="elements"))
    else:
        for g in S4:
            left_set = {g * h for h in sub}
            right_set = {h * g for h in sub}
            if left_set != right_set:
                print("Контрприклад a =", g)
                print("aH =", left_set)
                print("Ha =", right_set)
                break
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
from sage.libs.gap.libgap import libgap

K = GF(3)
GL23 = GL(2, K)
S4 = SymmetricGroup(4)

reps = [vector(K, [1, 0]), vector(K, [0, 1]),
        vector(K, [1, 1]), vector(K, [1, 2])]

# вектор -> номер прямої
which_line = {}
for num, v in enumerate(reps, 1):
    which_line[tuple(v)] = num
    which_line[tuple(-v)] = num

def phi(A):
    return S4([which_line[tuple(v * A.matrix())] for v in reps])

all_matrices = list(GL23)
print("Порядок GL(2,3):", GL23.order())

print("\nДія твірних на прямих:")
for g in GL23.gens():
    print(g.matrix())
    print("  phi =", phi(g))

ok = all(phi(A * B) == phi(A) * phi(B) for A in all_matrices for B in all_matrices)
print("\nphi — гомоморфізм (перевірено на всіх парах):", ok)

ker = GL23.subgroup([A for A in all_matrices if phi(A) == S4.one()])
img = S4.subgroup([phi(A) for A in all_matrices])

print("\nПорядок ядра:", ker.order())
print("Матриці ядра:")
for A in ker:
    print(A.matrix())

print("Порядок образу:", img.order())
print("Образ = S4:", img == S4)

quot = libgap.FactorGroup(libgap(GL23), libgap(ker))
print("\nПорядок G/N:", quot.Size())
print("Структура G/N:", quot.StructureDescription())
print("G/N ≅ Im(phi):", libgap.IsomorphismGroups(quot, libgap(img)) != libgap.fail)
```

### 5*. Проекції групи $SL_2(\mathbb{Z})$ на $SL_2(\mathbb{Z}_n)$ 

Розглянемо спеціальну лінійну групу $SL_2(\mathbb{Z})$, яка породжується двома матрицями:
$$S = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix},\qquad T = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}. $$

1. Задайте гомоморфізм $\rho_n: SL_2(\mathbb{Z}) \to SL_2(\mathbb{Z}_n)$, який редукує матриці за модулем $n$. 
2. Для значень $n\in\{2, 3, \ldots, 10\}$ перевірте, чи є $\rho_n$ сюр'єктивним, виконавши такі кроки:<br>
    a. Побудуйте групу $G_n = SL_2(\mathbb{Z}_n)$.<br>
    b. Побудуйте групу $H_n = \langle \rho_n(S), \rho_n(T) \rangle$.<br>
    c. Порівняйте порядки груп $H_n$ та $G_n$.



```sage
# Ваш код тут:

S_mat = matrix(ZZ, [[0, -1],
                    [1,  0]])
T_mat = matrix(ZZ, [[1, 1],
                    [0, 1]])

def rho(M, n):
    return M.change_ring(Integers(n))

print("det S =", S_mat.det(), "; det T =", T_mat.det())

words = [S_mat * T_mat, T_mat * S_mat * T_mat, (S_mat * T_mat)^3, T_mat^5 * S_mat]
print("rho_n(XY) = rho_n(X)*rho_n(Y):",
      all(rho(X * Y, n) == rho(X, n) * rho(Y, n)
          for n in range(2, 11) for X in words for Y in words))

def generated_order(gens, n):
    R = Integers(n)
    e = identity_matrix(R, 2)
    e.set_immutable()
    seen = {e}
    frontier = [e]
    while frontier:
        nxt = []
        for X in frontier:
            for g in gens:
                Y = X * g
                Y.set_immutable()
                if Y not in seen:
                    seen.add(Y)
                    nxt.append(Y)
        frontier = nxt
    return len(seen)

def sl2_order_formula(n):
    return ZZ(n^3 * prod(1 - 1/p^2 for p in prime_divisors(n)))

print("\n n | |G_n| | формула | |H_n| | H_n = G_n")
print("-" * 44)

results = []
for n in range(2, 11):
    G_n = SL(2, Integers(n))                                 
    order_G = G_n.order()
    order_H = generated_order([rho(S_mat, n), rho(T_mat, n)], n)   
    equal = (order_H == order_G)                              
    results.append(equal)
    print("%2d | %5d | %7d | %5d | %s" %
          (n, order_G, sl2_order_formula(n), order_H, equal))

print("\nrho_n сюр'єктивний для всіх n = 2..10:", all(results))

print("\nВИСНОВКИ")
print("1. Для кожного n = 2..10 виконується |H_n| = |SL_2(Z_n)|, отже H_n = SL_2(Z_n).")
print("2. Гомоморфізм rho_n сюр'єктивний (гіпотеза: для всіх n >= 2).")
print("3. Ядро rho_n — головна конгруенц-підгрупа Gamma(n), а SL_2(Z)/Gamma(n) ≅ SL_2(Z_n).")
```
